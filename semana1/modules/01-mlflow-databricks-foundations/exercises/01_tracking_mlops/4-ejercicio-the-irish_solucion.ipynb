```python
import mlflow
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn import datasets
 
from sklearn.tree import DecisionTreeClassifier 
from sklearn.metrics import accuracy_score
```

```python
dataset = datasets.load_iris()
```

```python
X = dataset.data
y = dataset.target
X_train, X_test, y_train, y_test = train_test_split(X, y)
```

```python
mlflow.sklearn.autolog()
mlflow.set_experiment("/Users/carrascoalvarezgonzalo@gmail.com/4-ejercicio-the-irish-solucion")
```

# Manual search

```python
configs = [
    {'max_depth': 3, 'max_features': 2},
    {'max_depth': 5, 'max_features': 3},
    {'max_depth': 10, 'max_features': 4},
    {'max_depth': None, 'max_features': 'sqrt'},
]

results = []
for i, config in enumerate(configs):
    with mlflow.start_run(run_name=f"DT Config {i+1}"):
        dt = DecisionTreeClassifier(**config, random_state=42)
        dt.fit(X_train, y_train)

        accuracy = accuracy_score(y_test, dt.predict(X_test))
        results.append({"Config": config, "Accuracy": accuracy})
        print(f"Config {i+1}: {config} -> Accuracy: {accuracy:.4f}")

print(max(results, key=lambda x: x["Accuracy"]))
```

The best configuration is max_depth: None and max_features: sqrt, which obtained an accuracy of 0.921

# Grid Search

```python
with mlflow.start_run(run_name="Grid Search - Decision Tree"):

  # GridSearch for hyperparameter tuning
  from sklearn.model_selection import GridSearchCV

  param_grid = {
    'max_depth': [3, 5, 10, None],
    'max_features': [2, 3, 4],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
  }
  dt = DecisionTreeClassifier()
  clf = GridSearchCV(estimator=dt, param_grid=param_grid)
  clf.fit(X_train, y_train)
  print(clf.best_params_)
  print(clf.best_estimator_)

  # Predictions
  y_preds = clf.predict(X_test)
  acc = accuracy_score(y_test, y_preds)

  # Log test metrics
  mlflow.log_metric("test_accuracy", acc)

mlflow.end_run()
```

# Model comparison

```python
import pandas as pd

from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score, 
    precision_score, 
    recall_score, 
    f1_score,
    confusion_matrix,
    classification_report
)

modelos = {
    "Decision tree": DecisionTreeClassifier(max_depth=10, random_state=42),
    "Random forest": RandomForestClassifier(n_estimators=100, random_state=42),
    "Support vector machine": SVC(C=1.0, kernel="rbf", random_state=42),
    "KNN": KNeighborsClassifier(n_neighbors=5),
    "Logistic Regression": LogisticRegression(C=1.0, random_state=42)
}

mlflow.sklearn.autolog()

results = []
for name, model in modelos.items():
    with mlflow.start_run(run_name=f"{name} - Iris"):
        model.fit(X_train, y_train)
        y_preds = model.predict(X_test)

        acc = accuracy_score(y_test, y_preds)
        f1 = f1_score(y_test, y_preds, average="weighted")
        results.append({
            "Model": name,
            "Accuracy": acc,
            "F1-Score": f1
        })
        print(f"{name}: Accuracy = {acc:.4f}, F1 = {f1:.4f}")

# Visualize results
df = pd.DataFrame(results)
print(df.to_string(index=False))
```

# Error analysis

```python
import numpy as np

incorrect_indices = np.where(y_preds != y_test)[0]
 
print(f"Number of errors: {len(incorrect_indices)}")
print("Error analysis:")
for idx in incorrect_indices:
    true_class = dataset.target_names[y_test[idx]]
    pred_class = dataset.target_names[y_preds[idx]]
    print(f"  Sample {idx}: Real={true_class}, Prediction={pred_class}")
    print(f"    Attributes: {X_test[idx]}")
```

# Dimensionality reduction

```python
from sklearn.decomposition import PCA

# Reduce to 2 dimensions
pca = PCA(n_components=2)
# Fit only on the train set to prevent data leakage
X_train_pca = pca.fit_transform(X_train)
X_test_pca = pca.transform(X_test)

with mlflow.start_run(run_name=f"PCA - Decision Tree"):
    dt = DecisionTreeClassifier(max_depth=10, random_state=42)
    dt.fit(X_train_pca, y_train)
    y_preds_pca = dt.predict(X_test_pca)

    acc = accuracy_score(y_test, y_preds_pca)
    print(f"Accuracy: {acc}")

```
